## Amostras do método extrator de cores para extração de paleta de quadros do Portinari.

```python
import cv2
import os
from funcionalidades import cor_predominante
import matplotlib.pyplot as plt
```

```python
def rgbaiter(cor):
    b= cor[0]
    g= cor[1]
    r = cor[2]
    return [r,g,b]
```

```python

cores_nov = []
proporcoes_nov = []
nomes_nov = []

imagens_dir= '/home/lucca/Documents/Atividades escolares/impatech/Projeto portinari/Projeto-Portinari/id_Cores/Metodo_final/amostra'

# cria uma lista com todos os caminhos
cam_imagens = [f for f in os.listdir(imagens_dir) 
            if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif'))]


for idx,img in enumerate(cam_imagens):
    #juntando o nome do arquivo da imagem para o caminho do folder das imagens
    caminho_img = os.path.join(imagens_dir,img)

    imagem = cv2.imread(caminho_img)

    imagem = imagem.reshape((-1,3))


    cor,qtd = cor_predominante(imagem)
    cores_nov.append(cor)
    proporcoes_nov.append(qtd)
    nomes_nov.append(img)
```

### Retrato João candido 

<img src="14757_Retrato_de_Jo_o_Candido.jpeg" alt="retrato_joao" width="300">

```python
fig = plt.figure()
ax = fig.add_subplot()
print(f'O quadro com nome {nomes_nov[5]} teve cor pincipal com proporção de {proporcoes_nov[5]:.2f}%')
ax.imshow([list(map(rgbaiter,cores_nov[5]))])
ax.axis('off')
```
