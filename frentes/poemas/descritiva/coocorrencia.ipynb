```python
from collections import defaultdict, Counter
import re
import pandas as pd


#importando os csv necessários
poemas_df = pd.read_csv('/home/lucca/Documents/Atividades escolares/impatech/Projeto portinari/Projeto-Portinari/Data/Poemas/poemas.csv') 
```

```python
#limpando o csv rapido

poemas_df = poemas_df.loc[:,['id', 'estrofes', 'titulo', 'ano']]
```

```python

tabela_co_ocorrencia = defaultdict(Counter)
qtd_palavras = Counter()
aparicoes_por_poemas = Counter()

#contando a coocorência
for poema in poemas_df['estrofes']:
    # extrai as palavras e remove duplicatas dentro do mesmo poema
    palavras = list(set(re.findall(r'\b\w+\b', poema.lower())))

    #contagem de palavras por poema
    aparicoes_por_poemas.update(set(palavras))

    #contagem de palavras no total
    qtd_palavras += Counter(palavras)


    for palavra_alvo in palavras:
        for outra_palavra in palavras:
            if palavra_alvo != outra_palavra:
                # atualiza a contagem
                tabela_co_ocorrencia[palavra_alvo][outra_palavra] += 1
```

```python
df_palavras = pd.DataFrame(tabela_co_ocorrencia)
```

```python
for palavra in df_palavras.columns:
    df_palavras.loc[palavra,palavra] = qtd_palavras[palavra]
```

```python
#criando o dataframe generico para o merge de uma coluna
df = pd.DataFrame(aparicoes_por_poemas.items(), columns=['index', 'frequencia'])

#organizando o dataframe
df_palavras = df_palavras.reset_index()
```

```python
df_palavras = df_palavras.merge(df,how='left',on = 'index')
df_palavras.head(5)

df_palavras = df_palavras.fillna(0)
```

```python
df_palavras.to_csv('/home/lucca/Documents/Atividades escolares/impatech/Projeto portinari/Projeto-Portinari/Data/Poemas/Poemas_Palavras.csv',index=False)
```
